**ลบเฉพาะ endpoint `itcs355-6688012-lab5` (รวม staging) และ compute `lab5train`, `lab5release`**
หลังลบจะเรียก endpoint นี้ไม่ได้ และรันงานบนสอง compute นี้ไม่ได้จนกว่าจะสร้างใหม่

หากต้องการภาพ endpoint ที่ทำงานอยู่ ให้เก็บก่อนรัน cell นี้
ตรวจงานที่ยัง active และ ownership ก่อนลบทั้งสามรายการ; หากไม่ตรงจะหยุดก่อนลบ

แล้วส่ง SEND THIS SUMMARY กลับมา; รายงาน billing หลัง 24 ชั่วโมงยังต้องตรวจอีกครั้ง
นี่คือ runtime cleanup ไม่ใช่คำยืนยันว่าทั้ง Lab หรือค่าใช้จ่ายทั้งหมดเสร็จแล้ว


```python
from pathlib import Path
import platform
import subprocess
import sys

if platform.system() != "Linux" or "microsoft" not in platform.release().lower():
    raise RuntimeError("เลือก kernel itcs355-lab3 ฝั่ง WSL ก่อนครับ")
here = Path.cwd().resolve()
lab = next((base / "lab5" for base in (here, *here.parents)
            if (base / "lab5/scripts/cleanup_runtime.py").is_file()
            and (base / "lab4/README.md").is_file()), None)
if lab is None:
    raise RuntimeError("เปิด notebook จาก repository เดิมก่อนครับ")
print("Remove only the named Lab 5 endpoint and two tagged compute clusters; retain model/history.", flush=True)
with subprocess.Popen(
    [sys.executable, "-u", "-m", "scripts.cleanup_runtime"], cwd=lab,
    stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1,
) as task:
    try:
        for line in task.stdout:
            print(line, end="", flush=True)
        exit_code = task.wait()
    except KeyboardInterrupt:
        task.terminate()
        try:
            task.wait(timeout=10)
        except subprocess.TimeoutExpired:
            task.kill()
        raise
if exit_code:
    raise RuntimeError("Cleanup หยุด: ส่ง summary/error ด้านบนก่อน ยังไม่ถือว่าลบครบครับ")
print("ส่ง SEND THIS SUMMARY กลับมา แล้วหยุดตรงนี้ครับ", flush=True)
```
