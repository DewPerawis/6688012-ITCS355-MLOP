```python
from pathlib import Path
import platform
import subprocess
import sys

if platform.system() != "Linux" or "microsoft" not in platform.release().lower():
    raise RuntimeError("เลือก kernel itcs355-lab3 ฝั่ง WSL ก่อนครับ")
here = Path.cwd().resolve()
lab = next((base / "lab5" for base in (here, *here.parents)
            if (base / "lab5/scripts/finalize_batch.py").is_file()
            and (base / "lab4/README.md").is_file()), None)
if lab is None:
    raise RuntimeError("เปิด notebook จาก repository เดิมก่อนครับ")
print("One batch: checks + read-only inventory + fresh billing + consolidated reports.", flush=True)
with subprocess.Popen(
    [sys.executable, "-u", "-m", "scripts.finalize_batch"], cwd=lab,
    stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1,
) as task:
    try:
        for line in task.stdout:
            print(line, end="", flush=True)
        exit_code = task.wait()
    except KeyboardInterrupt:
        task.terminate()
        try:
            task.wait(timeout=10)
        except subprocess.TimeoutExpired:
            task.kill()
        raise
if exit_code:
    raise RuntimeError("Batch report หยุด: ส่ง error ด้านบน ไม่ต้องรัน parts เก่าซ้ำครับ")
print("ส่ง SEND THIS SUMMARY กลับมา แล้วหยุดตรงนี้ครับ", flush=True)
```
